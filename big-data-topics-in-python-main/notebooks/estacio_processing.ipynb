# Estácio — Mulheres formandas em cursos de TI

Tratamento dos dados de formandos dos cursos de TI da Estácio (campus Via Corpvs,
2023–2025), para a análise de participação feminina.

A planilha foi obtida via requerimento interno junto à instituição, de uso restrito
a este trabalho acadêmico.

```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()
while not (ROOT / 'data').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent

ESTACIO_FILE = ROOT / 'data' / 'raw' / 'estacio' / 'estacio_ti_2023_2025.xlsx'
PROCESSED    = ROOT / 'data' / 'processed' / 'estacio'
```

```python
df = pd.read_excel(ESTACIO_FILE)
print('rows:', len(df))
df.head()
```

## 1. Remoção de duplicatas

```python
df = df.drop_duplicates()
print('rows after removing duplicates:', len(df))
```

## 2. Participação feminina

```python
sex = df['TP_SEXO'].value_counts()
pct_f = sex.get('F', 0) / sex.sum() * 100
print(sex.to_dict())
print(f'%F = {pct_f:.1f}%  ({sex.get("F", 0)} women out of {sex.sum()} graduates)')
```

## 3. Mulheres por curso

```python
df_women = df[df['TP_SEXO'] == 'F'].copy()
df_women['CURSO'].value_counts()
```

## 4. Exportação

```python
df_women.to_csv(PROCESSED / 'Estacio_IT_Women.csv', index=False)
df_women.to_excel(PROCESSED / 'Estacio_IT_Women.xlsx', index=False)
print(f'{len(df_women)} women exported to data/processed/estacio/ (.csv and .xlsx)')
```
